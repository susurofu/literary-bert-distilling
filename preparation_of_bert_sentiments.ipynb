# Sentiment inference

```python
from sentiment_inference import OrdinalBertInference

import pickle
import pandas as pd
```

```python
hemingway_sentences = pd.read_json('/home/artem/Documents/reasoning_sentiment/data_hemingway/old_man_and_the_sea_sentences.json')
hemingway_sentences = list(hemingway_sentences[0])
```

## Large

```python
inference = OrdinalBertInference(
    model_path="models/bert_ordinal_large_final",
    batch_size=32,
    max_len=256
)

labels = inference.predict(
    hemingway_sentences,
    output_file="bert_sentimnets/large.json"
)
```
