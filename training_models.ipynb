# Training Models

```python
from model_trainer import ModelTrainer
```

## Large

```python
trainer = ModelTrainer(
    base_model="google-bert/bert-large-uncased",
    dataset="guten_sentiment.csv",
    num_classes=10,
    max_len=256,
    output_dir="./bert_ordinal/large",
    seed=1111
)
```

```python
df = trainer.read_validate_data()
print(df.head())
```

```python
trainer.set_training_parameters()
```

```python
trainer.tokenize_dataset()
```

```python
trainer.train()
```

```python
results = trainer.test_results

print("Test MAE:", results["test_mae"])
print("Test RMSE:", results["test_rmse"])
print("Test Accuracy:", results["test_accuracy"])
```

```python
checkpoints = trainer.evaluate_interim_models()
print(checkpoints)

trainer.save_best_models(
    best_model_dir="./models/bert_ordinal_large_final"
)
```
